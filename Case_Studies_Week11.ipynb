### **Case Study 1 - Sam**

**Participant Details:** <br> Sam is 22, autistic, and loves the rhythm of making coffee. He’s practised at home, knows his way around an espresso machine, and is ready to learn on the job. Busy peak-hour cafes overwhelm him, but a smaller independent cafe with a quieter morning shift would suit him well.

**Location Used:** <br>

**Transport Used:** <br> Driving, 15 minutes away

**AI Prompt Used:** <br> Autistic and loves rhythm of making coffee. Busy peak-hour cafes overwhelm him, a smaller independent cafe with a quieter morning shift would suit him well.

**Dashboard Results:**

cs2.PNG

**Overview:** <br>
The AI prompt used resulted in the "Sensory Sensitivity (Quiet)" advanced filter being toggled, which returned very few business results. For the analysis, we will consider the results without this filter selected, as most commercial Google Maps listings do not explicitly fill in optional attributes like noise levels, limiting overall results when these filters are selected. After removing this filter, the dashboard returned a set of candidate businesses. Our system's precision was already well-established, so our primary objective for this case study was to verify the recall. We checked this by replicating the search on Google Maps using matching category, location, and radius proximity parameters. The query we used was "cafes", and we verified that our pipeline successfully captured approximately 90% of all relevant local businesses, which was a highly satisfactory benchmark.

From this case study, several key improvements were identified. The first is to constrain the Claude API to only have the ability to apply the "Wheelchair Accessible (Likely)" advanced filter, leaving other subjective or sparsely populated attributes like "Sensory Sensitivity (Quiet)" to be manually selected. Another important change is to modify the table display logic to prioritise "Direct categories" over "Related workplaces", ensuring the most relevant matches appear at the top of the table.

**Overview:** <br>
Similar to the previous two case studies, both precision and recall were strong. The lower result count relative to prior tests and other council areas was expected, given the new location coordinates and the lower business density of the Shire of Cardinia. While the Claude API interpreted the prompt well enough, this test revealed edge-case misclassifications. For example, a "Tennis club" was mapped as a "Related workplace" when it should have been categorised as a "Direct category".


The main improvement to make from this case study is that system prompts require fine-tuning to improve edge-case classification logic. This should also be done without over-indexing on specific keywords that could degrade accuracy across broader search queries (overfitting).
