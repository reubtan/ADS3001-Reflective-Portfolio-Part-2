# **Reflective Portfolio Week 9**

### **Overview and Context**

As part of the overall project, we needed an accurate dataset of local businesses across Greater Melbourne. Due to the cost and lack of efficiency with official Google APIs, we opted to use the *Google Maps Scraper by Compass* on Apify.

During initial testing on the free tier, we used a generic search term "Businesses" paired with suburb names. Seeing valid results, we proceeded to do a paid run across the City of Monash which would act as our sample. However, our sample validation only checked if the sample entries existed on Google Maps (precision) rather than manually benchmarking a specific high-density area on Google Maps to verify if any businesses were missed (recall).

At the start of this week, after scraping most target areas, we discovered significant data gaps: apart from a few exceptions, major employers, franchise networks, and local sports clubs were entirely absent. Missing these key job-providing entities along with likely many others compromised the dataset's reliability and accuracy, which will likely result in a full re-scrape, wasting significant time and financial resources.

### **Root Causes Breakdown Analysis**

**Method:** <br>
The team executed the extraction pipeline without conducting thorough research on the Apify scraper's configuration parameters. As a result, key advanced features were missed such as exact coordinate-based bounding polygons and unfiltered search terms with a zoom level control. These were left unexplored and we expect that optimising these settings will result in complete and reliable coverage. These settings will be integrated into next week's re-scraping sample.

<br>

**Preparation:** <br>
No thorough plan of the full data scraping process was created. Instead it was reactively done without a formal, end-to-end operational roadmap with milestones. As a result, we lacked standardised criteria to evaluate what constituted a "successful" versus "unsuccessful" sample prior to a full execution. Verification methods were improvised after data extraction which led us to approve a fundamentally flawed sample.

<br>

**Measurement:** <br>
Data verification focused solely on confirming whether extracted records existed in Google Maps (precision). The validation process failed to check for missing entities on Google Maps (recall). Consequently, significant data gaps went unnoticed during both the sample evaluation and full-scale runs, omitting many key businesses and entities that would be vital for identifying employer opportunities.

<br>

**Leadership:** <br>
The execution and validation of the scraping pipeline were mostly delegated to a single team member without structured oversight or scheduled review checkpoints. As project leader, maintaining unconditional trust without active check-ins was a critical oversight. This lack of hands-on monitoring resulted in an unstructured workflow and allowed critical data gaps to exist for an extended period which should've been my responsibility to ensure the process was structured and planned.

<br>

### **Corrective Actions**



**Optimised Settings Configuration:** <br>Replace text-based search locations with exact polygons using coordinates and also omit restrictive search terms like "Businesses". Combining unfiltered runs with tuned zoom levels should ensure full spatial coverage and uncover many businesses that were hidden before.

<br>

**Validation (Precision and Recall):**<br>Implement post-scraping validation gates by not only ensuring the businesses in the scrape results exist, but also if any are missing from Google Maps. This should be done for multiple samples in different areas before approving full production runs.

<br>

### **Future Prevention**

**Active Leadership & Oversight:**<br>Shift from passive delegation to active technical oversight. Ensure regular milestone check-ins and shared validation checks for critical project deliverables.

<br>

**Comprehensive Research:** <br> Prior to using new tools, APIs, etc., systematically review all available configuration settings and documentation to find optimal layouts and achieve optimal results. Avoid assuming default parameters are sufficient for most tools used online.

<br>

**Planning & Testing:** <br> Standardise a formal and structured preparation phase for future workflows in any field, incorporating edge-case planning and small-scale trial benchmarks before committing to full execution.
